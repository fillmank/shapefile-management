```python
import geopandas as gpd
import pandas as pd

# --- Read the data ---
# You can loop through your building files
buildings_list = []
for i in range(1, 6):
    # Replace with your actual file paths
    file_path = f"path/to/building_file_{i}.shp"
    gdf = gpd.read_file(file_path)
    buildings_list.append(gdf)

# --- Merge the building layers ---
merged_buildings = pd.concat(buildings_list, ignore_index=True)

# --- Read the state/county layer ---
state_counties = gpd.read_file("path/to/your_state_counties.shp")

# --- Filter to just your state (if not already filtered) ---
state_boundary = state_counties[state_counties['STATE_NAME'] == 'Your State Name']

# --- 1. Clip the buildings to the state boundary ---
# This is the overlay operation: intersection of buildings and state boundary
clipped_buildings = gpd.overlay(merged_buildings, state_boundary, how='intersection')

# --- 2. Perform the Spatial Join to get County info ---
# This adds the county name/ID to each building based on location
buildings_with_county = gpd.sjoin(clipped_buildings, state_counties, how='left', predicate='intersects')

# --- 3. Dissolve by County to group buildings ---
# This will combine all building geometries in the same county
# IMPORTANT: Replace 'COUNTY_NAME' with the actual field name for counties in your state layer.
buildings_by_county = buildings_with_county.dissolve(by='COUNTY_NAME', aggfunc='sum')

# --- 4. Export the final result ---
buildings_by_county.to_file("path/to/output/Buildings_By_County.shp")
```
